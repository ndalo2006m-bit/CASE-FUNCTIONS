```sql
---1.
create table  products (
    product_id STRING,
    product_name STRING,
    price STRING
);
insert into products values
('1','Laptop','1200.00'),
('2','Phone','800.00'),
('3','Keyboard','45.00'),
('4','Monitor','300.00'),
('5','Mouse','25.00');
select* from products;
---1.1.
select product_name,
       price,
       CASE 
       WHEN price > R1000 THEN 'Exppensive'
       when price between R100 and R1000 then 'Mid-range'
       else 'Budget'
       end as price_category
from products;

---2.
create table orders (
    order_id STRING,
    customer_name STRING,
    amount STRING
);
insert into orders values 
('1','Alice','150.00'),
('2','Bob','560.00'),
('3','Charlie','999.99'),
('4','Diana','45.50'),
('5','Ethan','1200.00);
select* from orders;

---2.1.
select customer_name,
       amount,
       case 
       when order >= 1000 then 'High value'
       when order between 500-999.99 then 'Medium value'
       else 'Low value'
       end as order_value_category
from order;

---3.
create table employees (
emp_id STRING,
emp_name STRING,
department STRING,
salary STRING 
);
insert into employees values 
('1','John','IT','85000'),
('2','Sara','HR','60000'),
('3','Mark','IT','75000'),
('4','Lucy','Finance','95000'),
('5','Tom','HR','55000');
select* from employees;

---3.1.
select emp_name,
       department,
       salary,
       case
       when department  = 'IT' and salary > 80000 then 'Senior IT'
       when department = 'HR' and salary > 55000 then 'Experienced HR'
       else 'Staff'
       end as position_level
from employees;

---4.
create table students (
student_id STRING,
stuudent_name STRING,
score STRING
);
insert into students values
('1','Anna','92'),
('2','Ben','76'),
('3','Cara','59'),
('4','David','83'),
('5','Ella','68');
select* from students;

---4.1.
select student_name,
       score,
       case
       when score >= 90 then 'A'
       when score between 80 and 89 then 'B'
       when score between 70 and 79 then 'C'
       when score between 60 and 69 then 'D'
       else 'F'
       end as grade
from students;

---5.
create table deliveries (
delivery_id STRING,
delivery_time_minutes STRING
);
insert into deliveries values 
('1','45'),
('2','80'),
('3','30'),
('4','65'),
('5','100');
select* from deliveries;

---5.1.
select deoivery_id,
delivery_time_minutes,
case 
when delivery_time_minutes <= 30 mins then 'Fast'
when delivery_time_minutes between 31 and 60 mins then 'On Time'
else 'Late'
end as performance 
from deliveries;

---6.
create table tickets (
ticket_id STRING,
issue_type STRING,
priority STRING
);
insert into tickets values 
('1','Login Issue','1'),
('2','Server Down','3'),
('3','Slow Down','2'),
('4','Email Error','2'),
('5','Password Reset','1');
select* from tickets;

---6.1.
select issue_type,
priority,
case 
when priority = 3 then 'High'
when priority = 2 then 'Medium'
else 'Low'
end as priority_label
from tickets;

---7.
create table attendence (
student_id STRING,
days_present STRING,
total_days STRING
);
insert into attendence values 
('1','45','50'),
('2','30','50'),
('3','48','50'),
('4','25','50'),
('5','50','50');
select* from attendence;

---7.1.
select student_id,
(days_present/total_days)* 100 as attendence_percentage,
case 
when (days_present/total_days)* 100 >= 90 then 'Excellent'
when (days_present/total_days)* 100 >= 75 then 'Good'
else 'Needs Improvement'
end as attendence_status
from attendence;

---8.
create table products_inventory (
product_id STRING,
stock_qty STRING,
);
insert into  products_inventory values 
('1','5'),
('2','0'),
('3','25'),
('4','10'),
('5','3');
select* from products_inventory;

---8.1.
select product_id,
stock_qty,
case 
when stock = 0 then 'Out of stock'
when stock is between 1 and 5 then 'Low stock'
else 'In stock'
end as stock_status
from products_inventory;

---9. 
create table classes (
class_id STRING,
subject STRING,
enrolled_students STRING
);
insert into classes values 
('1','Math','30'),
('2','English','25'),
('3','Science','15'),
('4','Art','5'),
('5','History','20');
select* from classes;

---9.1.
select subject,
enrolled_students,
case 
when enrolled_students >= 25 then 'Large'
when enrolled_studentsis between 10 and 24 then 'Medium'
else 'Small'
end as class_size_category
from classes;

---10.
create table payments (
payment_id STRING
amount STRING,
payment_method STRING,
);
insert into payments values 
('1','50.00','Card'),
('2','200.00','Cash'),
('3','150.00','Card'),
('4','75.00','Paypal'),
('5','300.00','Cash');
select* from payments;

---10.1.
select payment_id,
payment_method,
amount,
case 
when payment_method = 'Cash' and amount >= 200.00 then 'Eligible for discount'
else 'Not eleigible'
end as discount_eligibiity
from payments;
```
